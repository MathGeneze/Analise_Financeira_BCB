* **Contexto:** "O time de vendas precisa saber quais bandeiras de cartão estão dominando os estabelecimentos comerciais para decidirmos onde atacar."

* **Tarefa:** Usando os dados de `estabelecimentos_credenciados_por_cartoes.csv`:
  1. Crie uma **Tabela Dinâmica** filtrando apenas o último trimestre disponível (2026-03-31).
  2. Calcule o **Market Share %** de cada bandeira com base na coluna `qtd_estab_ativos`;
